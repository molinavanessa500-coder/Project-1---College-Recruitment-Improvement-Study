### Question 1 - From the “how did you hear about County College of Morris” question, what was the highest ‘yes’ percentage from the choices?

```python
hear_values = [63.58, 43.93, 43.93, 43.93, 42.20, 26.59, 23.70, 19.65, 19.08, 6.94, 5.78, 4.62, 3.47, 2.89] #values from results
hear_labels = ['Family Member of Friend', 'CCM Web Site', 'HS Counselor', 'HS Teacher', 'Current CCM Student', 'Community Event', 'CCM Alumni', 'Social Media', 'Other', 
'In-App Advertisemen', 'Billboard', 'Employer', 'Radio', 'Television'] #renamed categories to better understand the findings
plt.barh(hear_labels, hear_values) #used due to the large number of categories. It was easier to read and understand.
plt.title("The Percentage of Students who answered 'yes' to how they heard about County College of Morris")
plt.xlabel("The Percentage")
plt.ylabel("How Students Heard About County College of Morris")
```

### 1 Analysis - Using a horizontal bar chart displays the percentage of students who have heard about CCM from friends and family (63.58%), CCM website (43.93%), high school counselor (43.93%) and high school teacher (43.93%). This is just the top 4 that are displayed on the bottom of the chart. It also shows us which advertisments we can move away from such as the television, billboard, and the radio. Providing more funding and focusing on advertising on ways more towards the end of the list can increase awareness of the college.

### Question 2 - What is the percentage of students that answered “some impact” or “high impact” in school-based environments (i.e. Middle/High school computing class)

```python
#used a horizontal bar chart because it is easier to read and compare the categories. Renamed the categories in impact_labels to make it easier to read.
impact_values = [48.55, 32.37, 29.48, 18.50, 16.76, 15.61, 12.14, 7.51]
impact_labels = ['Computing Class','HS Teacher or Counselor', 'Computing Competitions','Computing Club','AP Computing Class', 'Dual_Enrollment','Afterschool Computing Program','Summer Computing Program']
plt.barh(impact_labels, impact_values)
plt.title("Percentage of Students who answered 'some impact' or 'high impact' in school-based environments")
plt.ylabel("School-Based Environments")
plt.xlabel("Percentage of Some and High Impact Reported")
```

### Question 3 - What were the top 3 motivations of students obtaining a computing degree/certification?

```python
#A horizontal bar chart looked the best to showcase the top 4 motivators. Adding the title and x and y labels also made the chart clearer
motivation_labels = ['Get a Job in the Computing Field', 'Personal Enrichment', 'Professional Development', 'Career Advancement']
motivation_values = [96, 85, 81, 81]
plt.barh(motivation_labels, motivation_values)
plt.title("The Top 4 Motivators of Students Obtaining A Computing Degree/Certification")
plt.xlabel("Number of Students who Selected 'Yes' to the Motivation")
plt.ylabel("Motivation")
```

### Question 4 - What gender has the lowest percentage of enrollment?

```python
gender_labels = ['Woman', 'Man', 'Non-binary', 'I do not identify', 'Prefer not to say']
gender_values = [20.81, 73.41, 3.47, 0, 2.31]
plt.figure(figsize=(10, 10)) #to change figure size
plt.bar(gender_labels, gender_values)
plt.title("The Percentage of Each Gender Who Took the Survey")
plt.xlabel("Gender")
plt.ylabel("Percentage")
plt.grid(axis='y', linestyle='--') #added to make the chart better to understand.
```

### 4 Analysis -From this bar chart, we can see clearly that the majority of students who completed this survey are men with a percentage of 73.41. From this report, 20.81% who completed this survey are women, 2.31% prefer not to say, and 3.47% are non-binary. Those that do not identify were 0%. What we can say from this chart is that our current demographic are men which could also mean that our current advertisement strategies are targeting men. To increase the number of students working towards obtaining a computing degree/certification as well as overall enrollment, we can use create new marketing strategies targeting the those the identify as woman and non-binary. We can also advertise those that do not identify as a gender and those who prefer not to state their gender. Not only would this increase the number of computing majors and computing certifications, this could also help diversify the IT department which can lead to the sharing of ideas and experiences.

### Question 5- What are the percentages of people who chose not interested (1 & 2), somewhat interested (3), and interested (4 and 5). 

```python
interest_labels = ['Not Interested', 'Somewhat Interested', 'Interested']
interest_values = [23.73, 35.59, 40.58]
plt.pie(interest_values, labels=interest_labels, autopct='%1.1f%%') #autopct added to display the percentage inside the pie slice so that the viewer knowns the percentage from the chart.
plt.title("The percentage of non-computing students who are not, somewhat, and interested in taking more computing classes")
```

### 5 Analysis -From this pie chart we can see that less than half of non-computing students are interested in taking more computing classes at CCM. To increase enrollment of computer courses, we can focus on the 40.6% that are interested as well as the 35.6% of students who are teetering between not interested and interested. Because they are in the middle, we can look back at the motivation data and see if the top 4 motivators can influence their decision to continue taking computing classes.
